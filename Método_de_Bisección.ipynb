# METODO DE BISECCIÓN

Este método se usa para resolver un problema en la busqueda de las raices de ecuaciones no lineales.

Queremos resolver la ecuación $x^3 - 4x^2-10 = 0$ en el intervalo [1,2].

Para aplicar el método de bisección verificamos que la función f(x) evaluada en los extremos opuestos del intervalo [a,b] tiene signos opuestos, es decir, $f(a)*f(b)<0$.

El teorema del valor intermedio nos asegura que existe un número p tal que p esta en [a,b] tal que es una raiz de $f(x)$, es decir, $f(p)=0$.
Aproximaremos la solución calculando de manera iterativa el punto medio del intervalo:
                                                        $$p_i=\frac{b_i-a_i}{2}$$
Importamos la libreria numpy,que tiene funciones matemáticas y matplotlib,que permite graficar.

```python
from math import *
import numpy as np
import matplotlib.pyplot as plt
```

```python
def funcion1(x):
    return x**3 + 4*(x**2) - 10
```

```python
#Ingreso datos de entrada para los diferentes métodos a trabajar
a = 1
b = 2

#Guarda los valores inciales

a0 = a
b0 = b

#Guarda valorws inicales del error y del número de iteraciones
tol = 10**-5 #float(input("Ingrese el valor de la tolerancia: "))
nmax = 100 #float(input(""Ingrese el número máximo de iteraciones: ))
error = 100
niter = 1
```

```python
#Método de Biseccion
#Evaluo el primer valor medio
m = a + (b - a)/2
niter = 1

#Evaluacion de los puntos en a,b y m
fa = funcion1(a)
fb = funcion1(b)
fm = funcion1(m)

#Se manda a imprimir las columnas de la tabla para las iteraciones
print(f"# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")
print(f"{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter, a0, fa, b0, fb, m, fm, error))

#Ciclo iterativo
while niter <= nmax:
    
    m = a + (b - a) / 2
    fm = funcion1(m)
    
    error = abs(m - a) 
    
    print(f"{niter} \t\t {a:.6f} \t {fa:.6f} \t {b:.6f} \t {fb:.6f} \t {m:.6f} \t {fm:.6f} \t {error:.6f}")
    
    if fm == 0 or (b - a) / 2 < tol:
        print(f"\nProcedimiento completado exitosamente. La raíz aproximada es: {m:.7f}")
        break # Sale del ciclo while
        
    niter += 1
    
    if fa * fm > 0:
        a = m
        fa = fm
    else:
        b = m
        fb = fm

else:
    print(f"\nEl método fracasó después de {nmax} iteraciones.")
```

```python
#Vamos a graficar la función 
xx = np.linspace(-5,2,100)
plt.plot(xx,funcion1(xx))
plt.title("Gráfica $x^3-4x^2-10$")
plt.xlabel("$x$")
plt.ylabel("$f(x)$")
plt.show()
```
