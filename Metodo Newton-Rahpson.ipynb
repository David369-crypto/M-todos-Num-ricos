El método de Newton-Raphson es un método numérico iterativo que permite aproximar las raíces de una función utilizando su derivada. Consiste en partir una aproximación lineal y calcular nuevas aproximaciones mediante la ecuación: 
$$p_n = p_{n-1} - \frac{f(p_{n})}{f'(p_{n-1})},\qquad n \geq 1$$

Se usará el código para encontrar las iteraciones de la siguiente función: 
$$f(x)=cosx-x$$

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
import pandas as pd
```

```python
def f(x):
      return np.cos(x) - x
```

```python
def df(x):
    return -np.sin(x) -1
```

```python
xx = np.linspace(0,1.2,100)
plt.plot(xx,f(xx))
plt.axhline(y=0,color="black")
plt.show()
```

```python
def newton(p0,n):
    lista = []
    error = []

    for i in range(n):
        fp0 = f(p0)
        dif = df(p0)

        p = p0 - (fp0/dif)
        lista.append(p)
        e = np.abs(p - p0)
        error.append(e)
        p0 = p
        
    return lista,error
```

```python
aproximaciones, errores = newton(np.pi/4, 4)

# Mostrar resultados en una tabla
tabla = pd.DataFrame({
    "Iteración": range(1, 5),
    "Aproximación": aproximaciones,
    "Error": errores
})

print(tabla)
```

La raiz aproximada de la función es: $x=0.739085$
